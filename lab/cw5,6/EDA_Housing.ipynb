## 5. Obsługa wartości odstających

Outliery = skrajne wartości. Wykrywanie: percentyle / **IQR** / odchylenie std.

Dla houseprice biorę 2 zmienne najbardziej „podejrzane”: **SalePrice** i **LotArea**.


### SalePrice — metoda IQR


```python
p0 = df.SalePrice.min()
p100 = df.SalePrice.max()
q1 = df.SalePrice.quantile(0.25)
q2 = df.SalePrice.quantile(0.5)
q3 = df.SalePrice.quantile(0.75)
iqr = q3 - q1

lc = q1 - 1.5 * iqr
uc = q3 + 1.5 * iqr

print('p0 =', p0, ', p100 =', p100, ', lc =', lc, ', uc =', uc)
print('q1 =', q1, ', q2 =', q2, ', q3 =', q3, ', iqr =', iqr)
```

**Konwencja:**
- jeśli `lc < p0` → brak outlierów od dołu
- jeśli `uc > p100` → brak outlierów od góry


```python
print('Outliery od dołu (SalePrice):', lc >= p0)
print('Outliery od góry (SalePrice):', uc < p100)
```

```python
df.SalePrice.plot(kind='box')
plt.title('SalePrice — przed przycięciem')
plt.show()
```

### LotArea — metoda IQR


```python
p0_la = df.LotArea.min()
p100_la = df.LotArea.max()
q1_la = df.LotArea.quantile(0.25)
q2_la = df.LotArea.quantile(0.5)
q3_la = df.LotArea.quantile(0.75)
iqr_la = q3_la - q1_la

lc_la = q1_la - 1.5 * iqr_la
uc_la = q3_la + 1.5 * iqr_la

print('p0 =', p0_la, ', p100 =', p100_la, ', lc =', lc_la, ', uc =', uc_la)
print('q1 =', q1_la, ', q2 =', q2_la, ', q3 =', q3_la, ', iqr =', iqr_la)
print('Outliery od dołu (LotArea):', lc_la >= p0_la)
print('Outliery od góry (LotArea):', uc_la < p100_la)
```

```python
df.LotArea.plot(kind='box')
plt.title('LotArea — przed przycięciem')
plt.show()
```

### Leczenie outlierów (clipping)

Nie usuwamy wierszy — przycinamy wartości poza `lc`/`uc`.


```python
# SalePrice — głównie górne outliery
df['SalePrice'] = df['SalePrice'].clip(upper=uc)
df.SalePrice.plot(kind='box')
plt.title('SalePrice — po przycięciu')
plt.show()
```

```python
# LotArea — outliery głównie z góry (też trochę od dołu)
df['LotArea'] = df['LotArea'].clip(lower=lc_la, upper=uc_la)
df.LotArea.plot(kind='box')
plt.title('LotArea — po przycięciu')
plt.show()
```

## 6. Obsługa wartości brakujących

Wykrywanie + leczenie.

W houseprice: zmienne z brakami **> 50%** usuwamy, poniżej 50% — imputujemy
(float → średnia, pozostałe → moda).


```python
df.isna()
```

### Procent braków w każdej zmiennej


```python
missing = df.isna().sum()
missing_pct = (missing / df.shape[0] * 100).sort_values(ascending=False)
braki = pd.DataFrame({'braki': missing, 'procent': missing_pct})
braki[braki['braki'] > 0]
```

```python
do_usuniecia = missing_pct[missing_pct > 50].index.tolist()
do_imputacji = missing_pct[(missing_pct > 0) & (missing_pct <= 50)].index.tolist()

print('Do usunięcia (>50%):', do_usuniecia)
print('Do imputacji (<=50%):', do_imputacji)
```

### Imputacja braków ≤ 50%


```python
for col in do_imputacji:
    if df[col].dtype == 'float64' or df[col].dtype == 'float32':
        wartosc = df[col].mean()
        df[col] = df[col].fillna(wartosc)
        print(f'{col} (float) → mean = {wartosc:.4f}')
    else:
        # object / kategorie — moda (top z describe)
        wartosc = df[col].mode()[0]
        df[col] = df[col].fillna(wartosc)
        print(f'{col} → mode = {wartosc}')
```

### Usunięcie zmiennych z brakami > 50%


```python
df.drop(columns=do_usuniecia, inplace=True)
print('Usunięto:', do_usuniecia)
print('Pozostało kolumn:', df.shape[1])
print('Suma braków po leczeniu:', df.isna().sum().sum())
```
